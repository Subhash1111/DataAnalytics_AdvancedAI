## 1. What is Deep Learning?
Deep Learning is a subset of Machine Learning that uses **multi-layer neural networks** to automatically learn representations from data.

**Key idea:** Feature learning is automatic.

## 2. Neural Networks – Intuition
- Inspired by the human brain
- Layers of neurons transform input data
- Deeper layers learn abstract features

**Example:** Image → edges → shapes → objects

## 3. Tensors
A tensor is a multi-dimensional array.
- Scalar → 0D
- Vector → 1D
- Matrix → 2D
- Image → 3D


## 4. Training Lifecycle
1. Input data
2. Forward propagation
3. Loss calculation
4. Backpropagation
5. Weight update
6. Repeat until convergence

## 5. Loss Functions
- Measure how wrong the model is
- Examples:
  - Mean Squared Error (Regression)
  - Cross Entropy Loss (Classification)


## 6. Optimizers
Optimizers update model weights.
- SGD
- Adam (most popular)
- RMSProp


## 7. Overfitting
- Model memorizes training data
- Performs poorly on new data

**Solutions:** Regularization, Dropout, More data

## 8. Evaluation Metrics
- Accuracy
- Precision / Recall
- F1 Score


## 9. Business Perspective
Deep Learning is best used when:
- Data is unstructured
- Accuracy matters more than explainability
- Large-scale automation is required


## 10. Summary
- Deep Learning learns features automatically
- Requires large data and compute
- Powers modern AI systems
