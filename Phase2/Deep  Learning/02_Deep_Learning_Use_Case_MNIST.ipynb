## Define Neural Network

````python
class DigitClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(28*28, 128)
        self.fc2 = nn.Linear(128, 64)
        self.fc3 = nn.Linear(64, 10)

    def forward(self, x):
        x = x.view(-1, 28*28)
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        return self.fc3(x)
````

## Train Model

````python
for epoch in range(5):
    for images, labels in train_loader:
        preds = model(images)
        loss = loss_fn(preds, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f'Epoch {epoch+1}, Loss: {loss.item():.4f}')
````

## 🔁 What Is an Epoch? (Deep Learning Theory Explained Simply)

---

### 🔹 Definition (Simple)
An **epoch** means **one complete pass of the entire training dataset through the neural network**.

📌 If the dataset has 60,000 images and we train for 1 epoch, the model sees **all 60,000 images once**.

---

### 🔹 Why Do We Need Epochs?

A model **does not learn everything in one pass**.

- Initial predictions are mostly random
- The model makes mistakes
- It adjusts its weights slightly
- Repeating this process helps the model improve

📌 *Learning happens gradually, not instantly.*

---

### 🔹 Epoch vs Batch (Very Important)

| Term | Meaning |
|----|-------|
| **Epoch** | One full pass over the entire dataset |
| **Batch** | A small group of samples processed together |
| **Iteration** | One batch processed once |

📌 Example:
- Dataset size: 60,000 images  
- Batch size: 64  
- 1 epoch = ~938 batches (iterations)

---

### 🔹 What Happens During One Epoch?

During **one epoch**:
1. Data is divided into batches
2. Each batch is sent to the model
3. Predictions are made
4. Loss is calculated
5. Gradients are computed
6. Weights are updated
7. This repeats until all data is used

📌 *At the end of an epoch, the model has seen every training example once.*

---

### 🔹 Why Don’t We Train for Just 1 Epoch?

Because:
- One pass is not enough to learn patterns
- The model needs repeated exposure
- Complex patterns need more refinement

📌 *Just like humans need practice, models need multiple epochs.*

---

### 🔹 Too Few vs Too Many Epochs

| Situation | Result |
|---|---|
| Too few epochs | Model underfits (has not learned enough) |
| Optimal epochs | Good generalization |
| Too many epochs | Model overfits (memorizes training data) |

📌 *Finding the right number of epochs is a balance.*

---

### 🔹 How Do We Choose Number of Epochs?

Depends on:
- Dataset size
- Model complexity
- Learning rate
- Problem difficulty

Typical ranges:
- Small problems: 5–10 epochs
- Medium problems: 10–50 epochs
- Large models: 100+ epochs

---

### 🔹 Real-World Analogy (Best for Teaching)

> Studying one chapter once = 1 epoch  
> Studying the same chapter multiple times = multiple epochs  
> More study helps, but too much memorization can hurt understanding
