## 3. `is` vs `==` (Critical Concept)

- `==` checks **value equality**
- `is` checks **object identity** (memory location)

👉 In pandas/NumPy, **always use `==` for comparisons**,  
👉 Use `is` **only** for `None`


```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

a == c, a is c
```

```python
x = None
x is None
```

## 6. Control Flow

Control flow is foundational for:
- Data cleaning
- Rule-based feature creation


```python
x = -5

if x < 0:
    label = "negative"
elif x == 0:
    label = "zero"
else:
    label = "positive"

label
```

## 7. Loops

⚠️ In real-world data science:
- Prefer **vectorized operations**
- Loops are mainly for learning or edge cases


```python
values = [1, 2, None, 4, None, 5]

total = 0
for v in values:
    if v is None:
        continue
    total += v

total
```

```python
values = [1, 2, 0, 4, 6, 5, 2, 1]

running_sum = 0
for v in values:
    if v == 5:
        break
    running_sum += v

running_sum
```

## 8. Pythonic Patterns (Highly Recommended)

### List Comprehensions


```python
values = [1, 2, None, 4]
cleaned = [v for v in values if v is not None]
cleaned
```

### Built-in Functions

Common in data pipelines:
- `sum`
- `min`, `max`
- `len`


```python
# basic level

numbers = [1, 2, 3, 4, 5]

squares = []

for n in numbers:
    squares.append(n*n)

print(squares)
```

```python
# Pro Level

numbers = [1, 2, 3, 4, 5]

squares = [n * n for n in numbers]

print(squares)
```

```python
# Pro ++
def square(n: int) -> int:
    return n * n

numbers = [1, 2, 3, 4, 5]

squares = list(map(square, numbers))

print(squares)
```

```python
data = [1, 2, 3, 4]
sum(data), len(data), min(data), max(data)
```
