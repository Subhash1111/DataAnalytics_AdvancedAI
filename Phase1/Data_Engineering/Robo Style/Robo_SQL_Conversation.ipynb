
# 🤖 SQL Conversation — *Robo at the Salon*
This notebook formats the *Robo movie–style* conversation as clean course material and includes a tiny **SQLite** demo so learners can run the SQL that the "Robo" assistant would generate.



## 🎬 Scene: User wants to retrieve their phone number

> **User:** Can you get my phone number?  
> **Robo:** What is your name?

> **User:** Bhajanlal  
> **Robo (SQL Translation):**
```sql
SELECT phone_number
FROM user
WHERE name = 'Bhajanlal';
```
> **Robo (Response):** There are **12 Bhajanlals**. Please provide your **initial**.

> **User:** Bhajanlal K  
> **Robo (SQL Translation):**
```sql
SELECT phone_number
FROM user
WHERE name = 'Bhajanlal K';
```
> **Robo (Response):** There are **3 Bhajanlals** with that name. Please provide your **address**.

> **User:** 6 Mallareddy Street, Chennai  
> **Robo (SQL Translation):**
```sql
SELECT phone_number
FROM user
WHERE name = 'Bhajanlal K'
  AND address = '6 Mallareddy Street, Chennai';
```
> **Robo (Final Answer):** 📞 **24884030**



## 🧩 Key Learning Takeaways

| Concept | Explanation |
|---|---|
| **Incremental Query Refinement** | Each user input adds a new filter (`WHERE` condition). |
| **Disambiguation** | When multiple results exist, the assistant asks for more identifying info. |
| **SQL Representation** | Natural language can be translated into executable SQL. |
| **Conversation Context** | The assistant maintains state across turns. |

## 🧪 Hands-on Demo (SQLite)
Run the cells below to create a small `user` table with data that matches the conversation. Then execute the SQL translations to see the results.


````

import sqlite3, pandas as pd, os
from IPython.display import display

DB_PATH = "robo_demo.db"

# Fresh DB
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.executescript("""
DROP TABLE IF EXISTS user;
CREATE TABLE user(
    id INTEGER PRIMARY KEY,
    name TEXT,
    address TEXT,
    phone_number TEXT
);
""")

# Insert rows
rows = []

# 12 entries for name 'Bhajanlal' (some without initial)
for i in range(1, 13):
    rows.append((i, "Bhajanlal", f"Random Street #{i}, Chennai", f"24884{i:03d}"))

# Now specific 'Bhajanlal K' entries (3 of them), one matching the final address
rows.extend([
    (100, "Bhajanlal K", "6 Mallareddy Street, Chennai", "24884030"),   # final expected
    (101, "Bhajanlal K", "12 Ramnagar, Chennai", "24884111"),
    (102, "Bhajanlal K", "23 MG Road, Chennai", "24884222"),
])

cur.executemany("INSERT INTO user(id, name, address, phone_number) VALUES (?,?,?,?)", rows)
conn.commit()

print("Database prepared:", DB_PATH, "Total rows:", len(rows))
````


### Step 1 — User: *"Bhajanlal"*
**SQL Translation**
```sql
SELECT phone_number
FROM user
WHERE name = 'Bhajanlal';
```


````

import sqlite3, pandas as pd
conn = sqlite3.connect("robo_demo.db")
q1 = """
SELECT phone_number
FROM user
WHERE name = 'Bhajanlal';
"""
df1 = pd.read_sql_query(q1, conn)
print("Matches:", len(df1))
display(df1.head())  # show a few rows
conn.close()
````
